```python
import requests

BASE_URL = "http://localhost:8000"

# 1) Obtenir un token
# route est '/auth/login'
resp = requests.post(
    f"{BASE_URL}/auth/login",
    json={"username": "admin", "password": "admin"},
)

print("Status login:", resp.status_code)

if resp.status_code == 200:
    data = resp.json()
    token = data["access_token"]
    headers = {"Authorization": f"Bearer {token}"}
    print(f"Token récupéré : {token[:10]}...")
else:
    print("Erreur d'authentification:", resp.text)
```

```python
# 2) Stats d'engagement filtrées
resp = requests.get(
    f"{BASE_URL}/api/engagement_stats",
    params={
        "country": "France",
        "segment": "Fit Relaxed",
        "per_page": 5,
    },
    headers=headers,
)

print("Status /api/engagement_stats:", resp.status_code)

if resp.status_code == 200:
    print(resp.json())
else:
    print("Erreur:", resp.text)
```

```python
# 3) Top 3 recommandations (Utilisateurs cibles)
resp = requests.get(
    f"{BASE_URL}/api/recommendations", # route définie dans l'API
    params={
        "max_rank": 3,
        "segment": "Fit Relaxed"      # filtrer par 'segment'
    },
    headers=headers,
)

print("Status /api/recommendations:", resp.status_code)

if resp.status_code == 200:
    print(resp.json())
else:
    print("Erreur:", resp.text)
```

```python
# 4) Santé des utilisateurs (Churn & Bien-être)
# route EST protégée (besoin de headers)
resp = requests.get(
    f"{BASE_URL}/api/user_health",
    headers=headers
)

print("Status /api/user_health:", resp.status_code)

if resp.status_code == 200:
    data = resp.json()
    print("Aperçu des données:", data["data"][:2])
else:
    print("Erreur:", resp.text)
```
